```python
import main
from data import datasets
import plot
```

```python
W_binary, y = datasets.load_dataset("MNIST_4_9")
```

```python
history_poisson_signed_0 = main.Graph(W_binary, y).RunActiveLearning(
        classifier="Poisson",
        variant="signed",
        init_label_per_class=5,
        label_per_iter=10,
        num_iter=75,
        add_edges_per_node=0
    )
```

```python
history_poisson_signed_3 = main.Graph(W_binary, y).RunActiveLearning(
        classifier="Poisson",
        variant="signed",
        init_label_per_class=5,
        label_per_iter=10,
        num_iter=75,
        add_edges_per_node=3,
    )
```

```python
history_poisson_signed_n= main.Graph(W_binary, y).RunActiveLearning(
        classifier="Poisson",
        variant="signed",
        init_label_per_class=5,
        label_per_iter=10,
        num_iter=75,
        add_edges_per_node=W_binary.shape[0]
    )
```

```python
history_poisson_unsigned = main.Graph(W_binary, y).RunActiveLearning(
        classifier="Poisson",
        variant="unsigned",
        init_label_per_class=5,
        label_per_iter=10,
        num_iter=75,
    )
```

```python
history_poisson_cut = main.Graph(W_binary, y).RunActiveLearning(
        classifier="Poisson",
        variant="cut",
        init_label_per_class=5,
        label_per_iter=10,
        num_iter=75,
    )
```

```python
plot.plot_variant_comparison(
    history_poisson_signed_0,
    history_poisson_signed_3,
    history_poisson_signed_n,
    history_poisson_unsigned,
    history_poisson_cut)
```
